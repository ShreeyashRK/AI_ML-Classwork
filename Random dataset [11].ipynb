```python
import pandas as pd
import numpy as np
import random

random.seed(42)
np.random.seed(42)

n = 200
customer_id = list(range(1, n + 1))
age = [random.randint(18, 65) for _ in range(n)]
gender = [random.choice(['Male', 'Female']) for _ in range(n)]
city = [random.choice(['Delhi', 'Mumbai', 'Bangalore', 'Chennai', 'Kolkata']) for _ in range(n)]
experience = [max(0, a - 18 - random.randint(0, 5)) for a in age]
income = [random.randint(20000, 120000) + exp * random.randint(500, 2000) for exp in experience]
score = [min(100, max(1, int(np.random.normal(50 + inc/5000, 15)))) for inc in income]
purchased = [1 if (inc > 60000 and sc > 50) or random.random() > 0.7 else 0 for inc, sc in zip(income, score)]

df = pd.DataFrame({'Customer_ID': customer_id,'Age': age,'Gender': gender,'City': city,'Experience_Years': experience,'Income': income,'Spending_Score': score,'Purchased': purchased})

for col in ['Income', 'Spending_Score']:
    for i in random.sample(range(n), 10):
        df.loc[i, col] = np.nan

print(df.head())
```
